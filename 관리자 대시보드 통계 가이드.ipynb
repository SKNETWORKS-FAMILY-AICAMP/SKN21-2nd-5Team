# 관리자 대시보드 통계 가이드


## 📋 개요 (Overview)

본 보고서는 호텔 예약 데이터를 기반으로 구축된관리자 대시보 관리자 대시보드의 핵심 지표(KPIs)와 분석 차트(A-1 ~ C-4)의 목적을 통합 정리한 문서입니다. 
분석은 **리스크 관리, 재무 성과, 운영 효율성** 세 가지 핵심 영역에 초점을 두고 작성되었습니다.

#### Section A: 취소 및 리스크 관리 분석 (Cancellation & Risk)

**목표:** 예약 취소의 근본적인 원인을 식별하고, 높은 취소 위험을 가진 예약을 예측하여 선제적인 리스크 관리 전략을 수립합니다.

| 차트  | 제목                          | 분석 목적                                                                                  | 의사 결정 활용                                                                                         |
| --- | --------------------------- | -------------------------------------------------------------------------------------- | ------------------------------------------------------------------------------------------------ |
| A-1 | 월별 취소율 추이 (호텔 유형 및 통합)      | 시계열 리스크 파악: 호텔 유형별로 취소율이 언제(월별) 최고점을 찍는지, 전체 통합 취소율과 비교하여 특정 호텔이 더 큰 위험을 안고 있는지 파악합니다. | 계절적 수요 감소/증가 시점에 맞춰 가격 조정 또는 취소 방지 마케팅을 계획합니다.                                                   |
| A-2 | 취소율 vs. 보증금 유형 및 ADR        | 고위험군 조합 식별: 보증금 유형과 **평균 요금 그룹(ADR)**의 조합 중 취소율이 가장 높은 리스크 그룹을 찾습니다.                   | 환불 불가(Non-Refund)' 보증금 정책을 강화하거나, 고위험군 ADR 구간의 예약에 대해 선결제 의무를 확대합니다.                             |
| A-3 | 리드타임 분포가 취소에 미치는 영향 (밀도 플롯) | 취소 패턴 분석: 취소된 예약과 미취소된 예약의 예약 선행 기간(Lead Time) 분포가 어떻게 다른지 밀도 형태로 비교합니다.               | 리드타임이 긴 구간(예: 100일 이상)에서 취소율 밀도가 높다면, 장기 예약 고객에 대한 **맞춤형 리스크 관리(예: 중간 확인 전화, 특별 혜택 제공)**를 시작합니다. |
|     |                             |                                                                                        |                                                                                                  |
#### Section B: 재무 및 수익 분석 (Revenue & Finance)

**목표:** 가격 책정 전략의 성과와 수익 기여도가 높은 채널을 식별하여 재무 성과를 극대화합니다.

| **차트**  | **차트 제목**                | **분석 목적**                                                                                                      | **의사 결정 활용 방안**                                                                    |
| ------- | ------------------------ | -------------------------------------------------------------------------------------------------------------- | ---------------------------------------------------------------------------------- |
| **B-1** | **월별 평균 ADR 추이** (호텔 비교) | **가격 전략 성과 측정:** 두 호텔의 **월별 평균 판매 단가(ADR)**의 추이를 비교하여, 계절적 요인에 따른 가격 책정 전략의 **성공 여부**와 **호텔 간 가격 경쟁력**을 평가합니다. | 경쟁 호텔 대비 가격이 낮게 책정된 시기를 파악하여 다음 분기의 **가격 인상/인하** 근거를 마련하고 **수익 관리 팀**에 피드백을 제공합니다. |
| **B-2** | **시장 채널별 수익 기여도**        | **수익 채널 우선순위 지정:** 어떤 **시장 채널(Market Segment)**이 호텔 전체 예상 순수익에 가장 큰 기여를 하는지 파악합니다.                             | 수익 기여도가 높은 채널에 대해 **프로모션 투자**를 늘리거나, 낮은 채널과의 **수수료 협상**을 재검토하여 영업 예산을 효율적으로 배분합니다. |
#### Section C: 운영 효율 및 자원 관리 (Operations & Resource)

**목표:** 프론트 데스크, 하우스키핑, 컨시어지 등의 운영 부하를 측정하고 객실 재고 및 인력 수요 패턴을 분석하여 운영 효율성을 높입니다.

| **차트**  | **차트 제목**                    | **분석 목적**                                                                                              | **의사 결정 활용 방안**                                                                                 |
| ------- | ---------------------------- | ------------------------------------------------------------------------------------------------------ | ----------------------------------------------------------------------------------------------- |
| **C-1** | **변경 예약 건수 분포**              | **프론트 데스크 부하 측정:** 예약 변경이 **몇 건** 발생하는지 빈도 분포를 파악하여, 프론트 데스크 직원의 **업무 집중도**와 **인력 소모량**을 가늠합니다.        | 변경 건수가 잦은 특정 예약 그룹을 식별하고, 예약 변경이 많은 시간대에 **프론트 데스크 인력을 증원**합니다.                                 |
| **C-2** | **특별 요청 건수 분포**              | **하우스키핑/컨시어지 부하 예측:** 특별 요청(`total_of_special_requests`)이 몇 건 발생했는지 분포를 파악하여 **추가적인 자원 소모** 규모를 파악합니다. | 특별 요청이 많은 예약이 집중된 날짜에 **컨시어지 및 하우스키핑 인력**을 유연하게 배정하고, 요청 건수가 많은 예약에 대한 **서비스 품질 관리**를 강화합니다.    |
| **C-3** | **주말 vs. 주중 숙박 비율 분포**       | **수요 패턴 및 인력 계획:** 주말과 주중 숙박 일수의 **중앙값과 분포**를 비교하여 호텔의 주된 수요 패턴(레저 vs. 비즈니스)을 확인합니다.                   | 주말 숙박이 길다면 **F&B/레저 시설** 인력을, 주중 숙박이 길다면 **객실 정리/회의 시설** 인력을 우선 배정하는 **수요 기반 인력 계획**을 수립합니다.    |
| **C-4** | **객실 타입 사용 현황** (상위 4개 + 기타) | **객실 재고 분석:** 가장 많이 예약되는 **상위 4개 객실 타입**의 비중을 확인하여, 현재 호텔이 **가장 수요가 높은 객실**을 충분히 확보하고 있는지 판단합니다.       | 수요가 높은 객실 타입에 대해 **높은 ADR**을 책정하고, 비중이 낮은 객실에 대한 **리노베이션 또는 타겟 마케팅 전략**을 검토하여 객실 포트폴리오를 최적화합니다. |
